```python
import numpy as np
import random
```

```python
def getArray():
  array_method=input("do you want to enter array elements manually?(yes/no) ")
  if array_method.lower()=='yes':
    arr1=list(input("enter the array elements separated by space ").split())
    return np.array(arr1,dtype=int)


  elif array_method.lower()=='no':
    range1=int(input("enter the start range of array elements "))
    range2=int(input("enter the end range of array elements "))
    size=int(input("enter the size of the 1D array (number of elements) "))
    arr=np.random.randint(range1,range2,size)
    print(arr)
    return arr

  else:
    print("invalid input")


```

```python
def Arithmetic_operations(arr1,arr2,op):
  if op in arith_list and arr1.size==arr2.size:
    print("result...")
    if op=="+":
      print(arr1+arr2)
    elif op=="-":
      print(arr1-arr2)
    elif op=="*":
      print(arr1*arr2)
    elif op=="/":
      print(arr1/arr2)
    elif op=="**":
      print(arr1**arr2)
  elif arr1.size!=arr2.size:
    print("enter same number of array elements")
  else:
    print("invalid operator")
```

```python
def getArrithmeticOperation():
  print(arith_list)
  print()
  operator=input("choose any operator in Arithmetic operations ")
  arr1=getArray()
  print()
  arr2=getArray()

  print()
  Arithmetic_operations(arr1,arr2,operator)
```

```python
def statistical_operations(arr,op):
  if op in stat_list:
    print("result...")
    if op=="mean":
      print(arr.mean())
    elif op=="median":
      print(np.median(arr))
    elif op=="std":
      print(arr.std())
    elif op=="var":
      print(arr.var())
    elif op=="min":
      print(arr.min())
    elif op=="max":
      print(arr.max())
    elif op=="sum":
      print(arr.sum())
    elif op=="product":
      print(np.product(arr))
    elif op=="shape":
      print(arr.shape)
    elif op=="size":
      print(arr.size)
    elif op=="data type":
      print(arr.dtype)
  else:
    print("invalid operator")
```

```python
def getStatisticalOperation():
  print(stat_list)
  print()
  operator=input("choose any operator in statistical operations ")
  arr=getArray()
  print()
  statistical_operations(arr,operator)
```

```python
def getPossibleShapes(arr,shape):
  possible_shapes=[]
  if shape=='2d':
    for i in range(1,arr.size+1):
      if arr.size%i==0:
        possible_shapes.append((i,arr.size//i))
  elif shape=='3d':
    for i in range(1,arr.size+1):
        for j in range(1,arr.size//i+1):
          k=arr.size//(i*j)
          if i*j*k==arr.size:
            possible_shapes.append((i,j,k))
  print(possible_shapes)
```

```python
def Array_manipulation(arr,op):
  if op in Array_manipulation_list:
    print("result calculating...")
    if op=="reshape":
      shape_dim=input("enter the new shape dimension (2d or 3d) ")
      getPossibleShapes(arr,shape_dim)
      if shape_dim=='2d':
        shape_input = input("enter any one of the possible above shapes as (rows,columns) (e.g., (3,10)): ")
        cleaned_input = shape_input.strip().replace('(', '').replace(')', '')
        shape1_str, shape2_str = cleaned_input.split(',')
        shape1 = int(shape1_str)
        shape2 = int(shape2_str)
        if shape1 * shape2 == arr.size:
          print(arr.reshape(shape1, shape2))
        else:
          print("invalid shape")

      elif shape_dim=='3d':
        shape_input = input("enter any one of the possible above shapes as (dim1,dim2,dim3) (e.g., (2,3,5)): ")
        cleaned_input = shape_input.strip().replace('(', '').replace(')', '')
        shape1_str, shape2_str, shape3_str = cleaned_input.split(',')
        shape1 = int(shape1_str)
        shape2 = int(shape2_str)
        shape3 = int(shape3_str)
        if shape1 * shape2 * shape3 == arr.size:
          print(arr.reshape(shape1, shape2, shape3))
        else:
          print("invalid shape")

    elif op=="sort":
      print(np.sort(arr))
    elif op=="reverse sort":
      print(np.sort(arr)[::-1])
    elif op=="unique value extraction":
      print(np.unique(arr))
    elif op=="get first element":
      print(arr[0])
    elif op=="get last element":
      print(arr[-1])
    elif op=="filter values greater than mean":
      print(arr[arr>arr.mean()])
  else:
    print("invalid operator")
```

```python
def getArrayManipulation():
  print(Array_manipulation_list)
  print()
  operator=input("choose any operator in Array manipulation ")
  arr=getArray()
  print()
  Array_manipulation(arr,operator)
```

```python
op_type_list=["Arithmetic operations","statistical operations","Array manipulation","Exit"]
arith_list=["+","-","*","/","**"]
stat_list=["mean","median","std","var","min","max","sum","product","shape","size","data type"]
Array_manipulation_list=["reshape","sort","reverse sort","unique value extraction","get first element","get last element","filter values greater than mean"]
```

```python
def getType():
  print("*"*30)
  print("types of available operations")
  print("*"*30)
  i=0
  while i < len(op_type_list):
    print(i+1,op_type_list[i])
    i+=1
  print("*"*30)


  op_type=input("choose the type of operation to perform ")
  return op_type
```

```python
def getOperation():
  while True:
    op_type = getType()

    if op_type in op_type_list or op_type in "1234":
      print()
      if op_type=="Arithmetic operations" or op_type=='1' :
        getArrithmeticOperation()
        print()


      elif op_type=="statistical operations"or op_type=='2':
        getStatisticalOperation()
        print()


      elif op_type=="Array manipulation" or op_type=='3':
        getArrayManipulation()
        print()


      elif op_type=='Exit' or op_type=='4':
        break

    else:
      print("invalid operation")
```
