```python

students = np.array(['Alice', 'Bob', 'Charlie', 'David', 'Eve'])
subjects = np.array(['Math', 'Science', 'History', 'Art'])

marks = np.array([
    [85, 90, 78, 92],
    [78, 88, 85, 90],
    [92, 76, 88, 85],
    [88, 85, 90, 78],
    [90, 92, 76, 88]
])


print(students)

print(subjects)

print(marks)
```

```python

print(students.ndim)
print(students.shape)
print(students.size)

print( subjects.ndim)
print(subjects.shape)
print(subjects.size)

print(marks.ndim)
print(marks.shape)
print(marks.size)
```

```python

marks_flat = marks.reshape(-1)

print(marks_flat)

marks_transposed = marks.T

print(marks_transposed)
```

```python
new_student_marks = np.array([[80, 88, 82, 95]])
# all_marks = np.vstack((marks, new_student_marks))
all_marks=np.concatenate((marks,new_student_marks))
print(all_marks)

new_subject_marks = np.array([[75], [80], [88], [85], [92], [89]])
new_subject_marks_reshaped = new_subject_marks.reshape(-1, 1)

print()

if all_marks.shape[0] == new_subject_marks_reshaped.shape[0]:
    all_marks_combined_with_subject = np.hstack((all_marks, new_subject_marks_reshaped))
    print(all_marks_combined_with_subject)

```
